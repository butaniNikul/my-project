```python
import random

""" 
1 for snake
-1 for water
0 for gun
"""
computer = random.choice([-1, 0, 1])
youstr = input("Enter your choice (s for snake, w for water, g for gun): ").lower()
youdict = {"s": 1, "w": -1, "g": 0}
reversedict = {1: "snake", -1: "water", 0: "gun"}

if youstr not in youdict:
    print("Invalid input! Please enter 's', 'w', or 'g'.")
else:
    you = youdict[youstr]

    print(f"You chose {reversedict[you]}\nComputer chose {reversedict[computer]}")

    if computer == you:
        print("It's a draw!")
    else:
        # computer=-1(water), you=1(snake) -> Snake drinks water -> You win
        if computer == -1 and you == 1:
            print("You win!")

        # computer=-1(water), you=0(gun) -> Water sinks gun -> You lose
        elif computer == -1 and you == 0:
            print("You lose!")

        # computer=1(snake), you=-1(water) -> Snake drinks water -> You lose
        elif computer == 1 and you == -1:
            print("You lose!")

        # computer=1(snake), you=0(gun) -> Gun shoots snake -> You win
        elif computer == 1 and you == 0:
            print("You win!")

        # computer=0(gun), you=-1(water) -> Water sinks gun -> You win
        elif computer == 0 and you == -1:
            print("You win!")

        # computer=0(gun), you=1(snake) -> Gun shoots snake -> You lose
        elif computer == 0 and you == 1:
            print("You lose!")
```
