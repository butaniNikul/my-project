# password generator

```python
import string
import random

def gen():
    s1 = string.ascii_uppercase # A-Z
  
    s2 = string.ascii_lowercase  # a-z
    
    s3 = string.digits  # 0-9
    
    s4 = string.punctuation # Special characters (!, @, #, $, etc.)
    
    passlen = int(input("enter the password length\n"))
    
    s = []
    s.extend(list(s1))  
    s.extend(list(s2))  
    s.extend(list(s3))
    s.extend(list(s4))
    
    random.shuffle(s)
    print(s)
    pas = ("".join(s[0:passlen]))
    print(pas)
    
gen()
```
